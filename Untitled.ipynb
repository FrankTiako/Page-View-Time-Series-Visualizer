```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from pandas.plotting import register_matplotlib_converters
```

```python
register_matplotlib_converters()
```

```python
df = pd.read_csv('fcc-forum-pageviews.csv')
```

```python
df['date'] = pd.to_datetime(df['date'])  
df = df.set_index('date')      
```

```python
df
```

```python
df = df[
    (df['value'] >= df['value'].quantile(0.025)) &
    (df['value'] <= df['value'].quantile(0.975))
]
```

```python
df
```

```python
fig1 = plt.figure(figsize=(14,4))
plt.plot(df.index, df['value'], color='red')
plt.title("Daily freeCodeCamp Forum Page Views 5/2016-12/2019")
plt.xlabel('Date')
plt.ylabel('Page Views')
```

```python
df_bar = df.copy()
df_bar["year"] = df_bar.index.year
df_bar["month"] = df_bar.index.strftime("%b")
df_grouped = df_bar.groupby(["year", "month"])["value"].mean()
```

```python
df_grouped = df_grouped.unstack()
months_order = ["January", "February", "March", "April", "May", "Jun",
                "July", "August", "September", "October", "November", "December"]

df_grouped = df_grouped[months_order]
```

```python
fig = df_grouped.plot(kind="bar", figsize=(10, 6))
plt.xlabel("Years")
plt.ylabel("Average Page Views")
plt.legend(title="Months")
plt.show()
```

```python
fig = df_grouped.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.xlabel("Years")
plt.ylabel("Average Page Views")
plt.legend(title="Months")
plt.tight_layout()

plt.show()
```

```python
df_box = df.copy()

# Préparer les données
df_box.reset_index(inplace=True)
df_box["year"] = df_box["date"].dt.year
df_box["month"] = df_box["date"].dt.strftime("%b")
df_box["month_num"] = df_box["date"].dt.month

# Trier les mois correctement
df_box = df_box.sort_values("month_num")

# Créer figure avec deux graphiques
fig, axes = plt.subplots(1, 2, figsize=(16, 5))

# 1️⃣ Year-wise Box Plot (Trend)
sns.boxplot(x="year", y="value", data=df_box, ax=axes[0])
axes[0].set_title("Year-wise Box Plot (Trend)")
axes[0].set_xlabel("Year")
axes[0].set_ylabel("Page Views")

# 2️⃣ Month-wise Box Plot (Seasonality)
sns.boxplot(x="month", y="value", data=df_box, ax=axes[1])
axes[1].set_title("Month-wise Box Plot (Seasonality)")
axes[1].set_xlabel("Month")
axes[1].set_ylabel("Page Views")
```
